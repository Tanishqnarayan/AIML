```python
# ============================================================
# AIML-6
# LINEAR REGRESSION
# ============================================================

# ============================================================
# 1. IMPORT LIBRARIES
# ============================================================

import os
import shutil
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.datasets import load_diabetes
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

from google.colab import files


# ============================================================
# 2. LOAD DIABETES DATASET
# ============================================================

data = load_diabetes(as_frame=True)

df = data.frame

print("=" * 60)
print("DATASET INFORMATION")
print("=" * 60)

print("Dataset shape:", df.shape)

print("\nFirst 5 rows:")
display(df.head())

print("\nColumns:")
print(df.columns.tolist())


# ============================================================
# 3. DEFINE X AND Y
# ============================================================

# Independent variables
X = df.drop(columns=["target"])

# Dependent variable
y = df["target"]

print("\n" + "=" * 60)
print("FEATURES AND TARGET")
print("=" * 60)

print("Number of independent features:", X.shape[1])
print("Number of samples:", X.shape[0])

print("\nIndependent features:")
print(X.columns.tolist())

print("\nDependent variable:")
print("target")


# ============================================================
# 4. TRAIN-TEST SPLIT
# ============================================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("\n" + "=" * 60)
print("TRAIN-TEST SPLIT")
print("=" * 60)

print("Training samples:", X_train.shape[0])
print("Testing samples:", X_test.shape[0])


# ============================================================
# 5. CREATE LINEAR REGRESSION MODEL
# ============================================================

model = LinearRegression()

# Train model
model.fit(
    X_train,
    y_train
)

print("\n" + "=" * 60)
print("MODEL TRAINING")
print("=" * 60)

print("Linear Regression model trained successfully.")


# ============================================================
# 6. MODEL COEFFICIENTS
# ============================================================

coefficients = pd.DataFrame({
    "Feature": X.columns,
    "Coefficient": model.coef_
})

print("\n" + "=" * 60)
print("MODEL COEFFICIENTS")
print("=" * 60)

display(
    coefficients
)


print("\nIntercept:")
print(model.intercept_)


# ============================================================
# 7. MAKE PREDICTIONS
# ============================================================

y_pred = model.predict(
    X_test
)

print("\n" + "=" * 60)
print("PREDICTIONS")
print("=" * 60)

print("First 10 predictions:")

prediction_preview = pd.DataFrame({
    "Actual": y_test.values[:10],
    "Predicted": y_pred[:10]
})

display(
    prediction_preview
)


# ============================================================
# 8. CALCULATE PERFORMANCE METRICS
# ============================================================

mae = mean_absolute_error(
    y_test,
    y_pred
)

mse = mean_squared_error(
    y_test,
    y_pred
)

rmse = np.sqrt(mse)

r2 = r2_score(
    y_test,
    y_pred
)

print("\n" + "=" * 60)
print("MODEL PERFORMANCE")
print("=" * 60)

print(
    f"Mean Absolute Error (MAE): {mae:.4f}"
)

print(
    f"Mean Squared Error (MSE): {mse:.4f}"
)

print(
    f"Root Mean Squared Error (RMSE): {rmse:.4f}"
)

print(
    f"R² Score: {r2:.4f}"
)

print(
    f"R² Score (%): {r2 * 100:.2f}%"
)


# ============================================================
# 9. ACTUAL VS PREDICTED VALUES
# ============================================================

comparison = pd.DataFrame({
    "Actual": y_test.values,
    "Predicted": y_pred,
    "Error": y_test.values - y_pred
})

print("\n" + "=" * 60)
print("ACTUAL VS PREDICTED")
print("=" * 60)

display(
    comparison.head(20)
)


# ============================================================
# 10. GRAPH 1 - ACTUAL VS PREDICTED
# ============================================================

plt.figure(figsize=(9, 6))

plt.scatter(
    y_test,
    y_pred,
    alpha=0.7
)

# Perfect prediction line
minimum = min(
    y_test.min(),
    y_pred.min()
)

maximum = max(
    y_test.max(),
    y_pred.max()
)

plt.plot(
    [minimum, maximum],
    [minimum, maximum],
    linestyle="--"
)

plt.xlabel(
    "Actual Target"
)

plt.ylabel(
    "Predicted Target"
)

plt.title(
    "Actual vs Predicted Values"
)

plt.grid(
    True,
    linestyle="--",
    alpha=0.5
)

plt.tight_layout()

plt.show()


# ============================================================
# 11. GRAPH 2 - RESIDUAL PLOT
# ============================================================

residuals = y_test - y_pred

plt.figure(figsize=(9, 6))

plt.scatter(
    y_pred,
    residuals,
    alpha=0.7
)

plt.axhline(
    y=0,
    linestyle="--"
)

plt.xlabel(
    "Predicted Values"
)

plt.ylabel(
    "Residuals"
)

plt.title(
    "Residual Plot"
)

plt.grid(
    True,
    linestyle="--",
    alpha=0.5
)

plt.tight_layout()

plt.show()


# ============================================================
# 12. GRAPH 3 - FEATURE COEFFICIENTS
# ============================================================

plt.figure(figsize=(10, 6))

plt.bar(
    coefficients["Feature"],
    coefficients["Coefficient"]
)

plt.xlabel(
    "Features"
)

plt.ylabel(
    "Coefficient"
)

plt.title(
    "Linear Regression Feature Coefficients"
)

plt.xticks(
    rotation=45
)

plt.grid(
    axis="y",
    linestyle="--",
    alpha=0.5
)

plt.tight_layout()

plt.show()


# ============================================================
# 13. GRAPH 4 - ACTUAL AND PREDICTED VALUES
# ============================================================

comparison_plot = comparison.reset_index()

comparison_plot = comparison_plot.head(30)

plt.figure(figsize=(12, 6))

plt.plot(
    comparison_plot["index"],
    comparison_plot["Actual"],
    marker="o",
    label="Actual"
)

plt.plot(
    comparison_plot["index"],
    comparison_plot["Predicted"],
    marker="x",
    label="Predicted"
)

plt.xlabel(
    "Test Sample"
)

plt.ylabel(
    "Target Value"
)

plt.title(
    "Actual vs Predicted Values for Test Samples"
)

plt.legend()

plt.grid(
    True,
    linestyle="--",
    alpha=0.5
)

plt.tight_layout()

plt.show()


# ============================================================
# 14. CREATE EXPORT FOLDER
# ============================================================

folder = "AIML_6_Linear_Regression_Results"

if os.path.exists(folder):
    shutil.rmtree(folder)

os.makedirs(folder)


# ============================================================
# 15. SAVE ORIGINAL DATASET
# ============================================================

df.to_csv(
    f"{folder}/diabetes_dataset.csv",
    index=False
)


# ============================================================
# 16. SAVE PREDICTIONS
# ============================================================

comparison.to_csv(
    f"{folder}/predictions.csv",
    index=False
)


# ============================================================
# 17. SAVE COEFFICIENTS
# ============================================================

coefficients.to_csv(
    f"{folder}/feature_coefficients.csv",
    index=False
)


# ============================================================
# 18. SAVE METRICS
# ============================================================

metrics_df = pd.DataFrame({
    "Metric": [
        "Mean Absolute Error (MAE)",
        "Mean Squared Error (MSE)",
        "Root Mean Squared Error (RMSE)",
        "R2 Score"
    ],
    "Value": [
        mae,
        mse,
        rmse,
        r2
    ]
})

metrics_df.to_csv(
    f"{folder}/model_metrics.csv",
    index=False
)


# ============================================================
# 19. SAVE GRAPH 1 - ACTUAL VS PREDICTED
# ============================================================

plt.figure(figsize=(9, 6))

plt.scatter(
    y_test,
    y_pred,
    alpha=0.7
)

plt.plot(
    [minimum, maximum],
    [minimum, maximum],
    linestyle="--"
)

plt.xlabel(
    "Actual Target"
)

plt.ylabel(
    "Predicted Target"
)

plt.title(
    "Actual vs Predicted Values"
)

plt.grid(
    True,
    linestyle="--",
    alpha=0.5
)

plt.tight_layout()

plt.savefig(
    f"{folder}/01_actual_vs_predicted.png",
    dpi=300,
    bbox_inches="tight"
)

plt.close()


# ============================================================
# 20. SAVE GRAPH 2 - RESIDUAL PLOT
# ============================================================

plt.figure(figsize=(9, 6))

plt.scatter(
    y_pred,
    residuals,
    alpha=0.7
)

plt.axhline(
    y=0,
    linestyle="--"
)

plt.xlabel(
    "Predicted Values"
)

plt.ylabel(
    "Residuals"
)

plt.title(
    "Residual Plot"
)

plt.grid(
    True,
    linestyle="--",
    alpha=0.5
)

plt.tight_layout()

plt.savefig(
    f"{folder}/02_residual_plot.png",
    dpi=300,
    bbox_inches="tight"
)

plt.close()


# ============================================================
# 21. SAVE GRAPH 3 - COEFFICIENTS
# ============================================================

plt.figure(figsize=(10, 6))

plt.bar(
    coefficients["Feature"],
    coefficients["Coefficient"]
)

plt.xlabel(
    "Features"
)

plt.ylabel(
    "Coefficient"
)

plt.title(
    "Linear Regression Feature Coefficients"
)

plt.xticks(
    rotation=45
)

plt.grid(
    axis="y",
    linestyle="--",
    alpha=0.5
)

plt.tight_layout()

plt.savefig(
    f"{folder}/03_feature_coefficients.png",
    dpi=300,
    bbox_inches="tight"
)

plt.close()


# ============================================================
# 22. SAVE GRAPH 4 - ACTUAL/PREDICTED TEST SAMPLES
# ============================================================

plt.figure(figsize=(12, 6))

plt.plot(
    comparison_plot["index"],
    comparison_plot["Actual"],
    marker="o",
    label="Actual"
)

plt.plot(
    comparison_plot["index"],
    comparison_plot["Predicted"],
    marker="x",
    label="Predicted"
)

plt.xlabel(
    "Test Sample"
)

plt.ylabel(
    "Target Value"
)

plt.title(
    "Actual vs Predicted Values for Test Samples"
)

plt.legend()

plt.grid(
    True,
    linestyle="--",
    alpha=0.5
)

plt.tight_layout()

plt.savefig(
    f"{folder}/04_test_actual_vs_predicted.png",
    dpi=300,
    bbox_inches="tight"
)

plt.close()


# ============================================================
# 23. SAVE OBSERVATIONS AND CONCLUSION
# ============================================================

observations = f"""
============================================================
AIML-6
LINEAR REGRESSION
============================================================

Dataset:
Scikit-learn Diabetes Dataset

Number of samples:
{df.shape[0]}

Number of independent features:
{X.shape[1]}

Training samples:
{X_train.shape[0]}

Testing samples:
{X_test.shape[0]}

============================================================
MODEL PERFORMANCE
============================================================

Mean Absolute Error (MAE):
{mae:.4f}

Mean Squared Error (MSE):
{mse:.4f}

Root Mean Squared Error (RMSE):
{rmse:.4f}

R² Score:
{r2:.4f}

R² Score (%):
{r2 * 100:.2f}%

============================================================
OBSERVATIONS
============================================================

1. Linear Regression was applied to predict the continuous
   target value of the Diabetes dataset.

2. The dataset contains 10 independent features and one
   continuous dependent variable.

3. The dataset was divided into training and testing sets
   using an 80:20 ratio.

4. The model coefficients indicate the contribution of
   each independent feature to the predicted target.

5. Mean Absolute Error, Mean Squared Error and Root Mean
   Squared Error were used to measure prediction error.

6. The R² score was used to measure the proportion of
   variation in the target explained by the model.

7. The Actual vs Predicted graph shows how closely the
   predicted values follow the actual target values.

8. The residual plot helps visualize the prediction errors.

============================================================
CONCLUSION
============================================================

Linear Regression was successfully implemented on the
Diabetes dataset. The model was trained using the ten
independent features to predict the continuous target
variable. Model performance was evaluated using MAE, MSE,
RMSE and R² score. Graphical analysis was also performed
using actual-versus-predicted and residual plots.

============================================================
"""

with open(
    f"{folder}/05_results_and_observations.txt",
    "w",
    encoding="utf-8"
) as f:

    f.write(observations)


# ============================================================
# 24. CREATE HTML REPORT
# ============================================================

html = f"""
<html>

<head>

<title>AIML-6 - Linear Regression Results</title>

<style>

body {{
    font-family: Arial;
    margin: 40px;
}}

h1, h2 {{
    color: #333;
}}

img {{
    width: 800px;
    max-width: 100%;
    margin-bottom: 30px;
}}

table {{
    border-collapse: collapse;
    width: 100%;
}}

th, td {{
    border: 1px solid #999;
    padding: 8px;
    text-align: center;
}}

th {{
    background-color: #eee;
}}

</style>

</head>

<body>

<h1>AIML-6: Linear Regression</h1>

<h2>Dataset Information</h2>

<p>
Dataset: Scikit-learn Diabetes Dataset
</p>

<p>
Samples: {df.shape[0]}
</p>

<p>
Independent Features: {X.shape[1]}
</p>

<p>
Training Samples: {X_train.shape[0]}
</p>

<p>
Testing Samples: {X_test.shape[0]}
</p>

<h2>Model Performance</h2>

<table>

<tr>
<th>Metric</th>
<th>Value</th>
</tr>

<tr>
<td>MAE</td>
<td>{mae:.4f}</td>
</tr>

<tr>
<td>MSE</td>
<td>{mse:.4f}</td>
</tr>

<tr>
<td>RMSE</td>
<td>{rmse:.4f}</td>
</tr>

<tr>
<td>R² Score</td>
<td>{r2:.4f}</td>
</tr>

</table>

<h2>Actual vs Predicted</h2>

<img src="01_actual_vs_predicted.png">

<h2>Residual Plot</h2>

<img src="02_residual_plot.png">

<h2>Feature Coefficients</h2>

<img src="03_feature_coefficients.png">

<h2>Test Sample Predictions</h2>

<img src="04_test_actual_vs_predicted.png">

<h2>Feature Coefficients Table</h2>

{coefficients.to_html(index=False)}

<h2>Conclusion</h2>

<p>
Linear Regression was successfully implemented on the
Diabetes dataset. The model was trained using ten
independent features to predict the continuous target.
Performance was evaluated using MAE, MSE, RMSE and R²
score.
</p>

</body>

</html>
"""

with open(
    f"{folder}/06_Linear_Regression_Complete_Report.html",
    "w",
    encoding="utf-8"
) as f:

    f.write(html)


# ============================================================
# 25. CREATE ZIP FILE
# ============================================================

zip_path = shutil.make_archive(
    "AIML_6_Linear_Regression_Complete_Results",
    "zip",
    folder
)


# ============================================================
# 26. DISPLAY EXPORTED FILES
# ============================================================

print("\n" + "=" * 60)
print("AIML-6 EXPORT COMPLETED")
print("=" * 60)

for filename in sorted(
    os.listdir(folder)
):

    print("✓", filename)

print("\nZIP file:")
print(zip_path)


# ============================================================
# 27. DOWNLOAD EVERYTHING
# ============================================================

files.download(zip_path)
```
