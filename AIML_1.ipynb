```python
from sklearn.datasets import load_diabetes
import pandas as pd

# Load the dataset
data = load_diabetes(as_frame=True)

# Convert into a DataFrame
df = data.frame

# Display the dataset
print(df.head())

# Check the dimensions
print("Dataset shape:", df.shape)

# Check column names
print(df.columns)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Independent variables
X = df.drop(columns=['target'])

# Dependent variable
y = df['target']

print("Independent variables:")
print(X.columns.tolist())

print("\nDependent variable:")
print(y.name)
```

```python
# Calculate correlation with the target
correlation = df.corr(numeric_only=True)['target']

# Remove the target's correlation with itself
correlation = correlation.drop('target')

# Sort correlations by absolute strength
correlation = correlation.reindex(
    correlation.abs().sort_values(ascending=False).index
)

print("Correlation with Diabetes Progression:")
print(correlation)
```

```python
plt.figure(figsize=(10, 6))

correlation.sort_values().plot(
    kind='barh',
    color='steelblue'
)

plt.title('Correlation of Independent Variables with Target')
plt.xlabel('Pearson Correlation Coefficient')
plt.ylabel('Independent Variables')
plt.axvline(0, color='black', linewidth=0.8)
plt.grid(axis='x', linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 9))

sns.heatmap(
    df.corr(numeric_only=True),
    annot=True,
    cmap='coolwarm',
    fmt='.2f',
    center=0
)

plt.title('Correlation Heatmap of Diabetes Dataset')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.regplot(
    data=df,
    x='bmi',
    y='target',
    scatter_kws={'alpha': 0.6},
    line_kws={'color': 'red'}
)

plt.title('BMI vs Diabetes Progression')
plt.xlabel('BMI (standardized)')
plt.ylabel('Diabetes Progression')
plt.tight_layout()
plt.show()
```

```python
# ==========================================
# ASSIGNMENT 1: EXPORT ALL RESULTS
# ==========================================

# 1. Import libraries
import os
import shutil
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import load_diabetes
from google.colab import files

# 2. Load dataset
data = load_diabetes(as_frame=True)
df = data.frame

# 3. Create folder for exported results
folder = "Assignment_1_Results"
os.makedirs(folder, exist_ok=True)

# 4. Calculate correlations
correlation = (
    df.corr(numeric_only=True)["target"]
    .drop("target")
    .sort_values(key=abs, ascending=False)
)

# Save correlation table
results = correlation.rename_axis(
    "Independent Variable"
).reset_index(name="Correlation")

results.to_csv(
    f"{folder}/correlation_results.csv",
    index=False
)

# Save the complete correlation matrix
df.corr(numeric_only=True).to_csv(
    f"{folder}/correlation_matrix.csv"
)

# ==========================================
# GRAPH 1: CORRELATION BAR CHART
# ==========================================

plt.figure(figsize=(10, 6))

correlation.sort_values().plot(
    kind="barh",
    color="steelblue"
)

plt.title("Correlation of Independent Variables with Target")
plt.xlabel("Pearson Correlation Coefficient")
plt.ylabel("Independent Variables")
plt.axvline(0, color="black", linewidth=0.8)
plt.grid(axis="x", linestyle="--", alpha=0.5)
plt.tight_layout()

plt.savefig(
    f"{folder}/correlation_bar_chart.png",
    dpi=300,
    bbox_inches="tight"
)
plt.show()
plt.close()

# ==========================================
# GRAPH 2: CORRELATION HEATMAP
# ==========================================

plt.figure(figsize=(12, 9))

sns.heatmap(
    df.corr(numeric_only=True),
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    center=0
)

plt.title("Correlation Heatmap - Diabetes Dataset")
plt.tight_layout()

plt.savefig(
    f"{folder}/correlation_heatmap.png",
    dpi=300,
    bbox_inches="tight"
)
plt.show()
plt.close()

# ==========================================
# GRAPH 3: BMI VS TARGET
# ==========================================

plt.figure(figsize=(8, 5))

sns.regplot(
    data=df,
    x="bmi",
    y="target",
    scatter_kws={"alpha": 0.6},
    line_kws={"color": "red"}
)

plt.title("BMI vs Diabetes Progression")
plt.xlabel("BMI (standardized)")
plt.ylabel("Diabetes Progression")
plt.tight_layout()

plt.savefig(
    f"{folder}/bmi_scatter_plot.png",
    dpi=300,
    bbox_inches="tight"
)
plt.show()
plt.close()

# ==========================================
# SAVE OBSERVATIONS
# ==========================================

strongest_feature = correlation.abs().idxmax()
strongest_value = correlation[strongest_feature]

observations = f"""
ASSIGNMENT 1
Relationship Between Dependent and Independent Variables

Dataset: Scikit-learn Diabetes Dataset
Number of samples: {df.shape[0]}
Number of independent variables: {len(correlation)}

Dependent variable: target

Correlation Results:
{results.to_string(index=False)}

Key Observation:
The feature with the strongest absolute linear correlation
with the target is '{strongest_feature}',
with a Pearson correlation coefficient of
{strongest_value:.4f}.

A positive coefficient indicates a positive linear
relationship, while a negative coefficient indicates
a negative linear relationship.

Conclusion:
Correlation analysis was used to examine the relationship
between independent variables and the target. Bar charts,
a correlation heatmap, and a scatter plot were generated.
Correlation indicates association, not causation.
"""

with open(
    f"{folder}/observations_and_conclusion.txt",
    "w",
    encoding="utf-8"
) as f:
    f.write(observations)

# ==========================================
# CREATE ZIP FILE
# ==========================================

zip_path = shutil.make_archive(
    "Assignment_1_Complete_Results",
    "zip",
    folder
)

print("Export completed successfully!")
print("Files included:")
for filename in sorted(os.listdir(folder)):
    print("-", filename)

# Download everything together
files.download(zip_path)
```
